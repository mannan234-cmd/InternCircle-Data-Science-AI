```python
# InternCircle - Data Science & AI Internship
# Task 01 - Data Lab Setup


import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("All required libraries imported successfully!")
```

```python
print("Pandas version:", pd.__version__)
print("NumPy version:", np.__version__)
print("Matplotlib version:", plt.matplotlib.__version__)
print("Seaborn version:", sns.__version__)
```

```python
# Test NumPy
numbers = np.array([10, 20, 30, 40, 50])

print("NumPy Array:")
print(numbers)
print("Mean:", numbers.mean())

# Test Pandas
data = {
    "Name": ["Ali", "Ahmed", "Sara", "Ayesha"],
    "Score": [85, 90, 78, 95]
}

df = pd.DataFrame(data)

print("\nPandas DataFrame:")
display(df)
```

```python
# Simple Data Visualization Test

plt.figure(figsize=(8, 5))

plt.bar(df["Name"], df["Score"])

plt.title("Sample Student Scores")
plt.xlabel("Student")
plt.ylabel("Score")

plt.show()
```

```python
# Seaborn Visualization Test

plt.figure(figsize=(8, 5))

sns.barplot(data=df, x="Name", y="Score")

plt.title("Student Score Analysis")

plt.show()
```
