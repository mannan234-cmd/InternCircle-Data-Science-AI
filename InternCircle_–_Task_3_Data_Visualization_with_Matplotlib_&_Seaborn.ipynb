## Data Visualization with Matplotlib & Seaborn

### Project: Titanic Dataset Visualization

This project focuses on visualizing the Titanic dataset using Matplotlib and Seaborn.

### Objectives
- Create informative bar charts and scatter plots.
- Create histograms to understand data distributions.
- Create box plots to identify distributions and outliers.
- Generate a correlation matrix.
- Create a correlation heatmap.
- Apply custom color palettes and plot formatting.
- Extract meaningful insights from visual data.

### Tools Used
- Python
- Pandas
- Matplotlib
- Seaborn
- Google Colab

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully!")
```

```python
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"

df = pd.read_csv(url)

print("Dataset loaded successfully!")
print("Shape of dataset:", df.shape)
```

```python
df.head()
```

```python
print("Dataset Information:")
df.info()
```

```python
print("Missing Values:")
print(df.isnull().sum())
```

```python
df["Age"] = df["Age"].fillna(df["Age"].median())
df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])

print("Missing values handled successfully!")
print(df.isnull().sum())
```

```python
# Set Seaborn style
sns.set_theme(style="whitegrid")

# Create a custom color palette
custom_palette = ["#3498DB", "#E74C3C", "#2ECC71", "#9B59B6"]

print("Visualization style and custom palette set successfully!")
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="Sex",
    hue="Survived",
    palette=["#E74C3C", "#2ECC71"]
)

plt.title("Titanic Survival by Gender", fontsize=16, fontweight="bold")
plt.xlabel("Gender")
plt.ylabel("Number of Passengers")
plt.legend(title="Survived", labels=["No", "Yes"])
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="Pclass",
    hue="Survived",
    palette=["#E74C3C", "#2ECC71"]
)

plt.title("Titanic Survival by Passenger Class", fontsize=16, fontweight="bold")
plt.xlabel("Passenger Class")
plt.ylabel("Number of Passengers")
plt.legend(title="Survived", labels=["No", "Yes"])
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x="Age",
    bins=20,
    kde=True
)

plt.title("Age Distribution of Titanic Passengers", fontsize=16, fontweight="bold")
plt.xlabel("Age")
plt.ylabel("Number of Passengers")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x="Fare",
    bins=30,
    kde=True
)

plt.title("Fare Distribution of Titanic Passengers", fontsize=16, fontweight="bold")
plt.xlabel("Fare")
plt.ylabel("Number of Passengers")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="Pclass",
    y="Age",
    hue="Pclass",
    palette="Set2",
    legend=False
)

plt.title("Age Distribution by Passenger Class", fontsize=16, fontweight="bold")
plt.xlabel("Passenger Class")
plt.ylabel("Age")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="Pclass",
    y="Fare",
    hue="Pclass",
    palette="Set3",
    legend=False
)

plt.title("Fare Distribution by Passenger Class", fontsize=16, fontweight="bold")
plt.xlabel("Passenger Class")
plt.ylabel("Fare")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(9, 6))

sns.scatterplot(
    data=df,
    x="Age",
    y="Fare",
    hue="Survived",
    palette=["#E74C3C", "#2ECC71"],
    alpha=0.7
)

plt.title("Relationship Between Age and Fare", fontsize=16, fontweight="bold")
plt.xlabel("Age")
plt.ylabel("Fare")
plt.legend(title="Survived", labels=["No", "Yes"])
plt.tight_layout()
plt.show()
```

```python
# Select numeric columns
numeric_data = df[["Survived", "Pclass", "Age", "SibSp", "Parch", "Fare"]]

# Create correlation matrix
correlation_matrix = numeric_data.corr()

print("Correlation Matrix:")
correlation_matrix
```

```python
plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)

plt.title("Correlation Heatmap of Titanic Dataset", fontsize=16, fontweight="bold")
plt.tight_layout()
plt.show()
```

```python
survival_correlation = correlation_matrix["Survived"].drop("Survived").sort_values()

plt.figure(figsize=(9, 5))

survival_correlation.plot(
    kind="bar",
    figsize=(9, 5)
)

plt.title("Correlation of Features with Survival", fontsize=16, fontweight="bold")
plt.xlabel("Features")
plt.ylabel("Correlation with Survival")
plt.xticks(rotation=45)
plt.axhline(0, linewidth=1)
plt.tight_layout()
plt.show()
```
